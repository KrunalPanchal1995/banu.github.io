## 1. Tokenizer: text to integer ids (Unit 3)
Predict first: will `'unbelievable'` be one token or several?

```python
from transformers import AutoTokenizer
tok = AutoTokenizer.from_pretrained('distilgpt2')
ids = tok('unbelievable tokenization')['input_ids']
print(ids); print(tok.convert_ids_to_tokens(ids))
```

## 2. Generate with a pipeline
Change `temperature` and `top_k` and compare with your Unit 3 decoding experiments.

```python
from transformers import pipeline, set_seed
set_seed(0)
gen = pipeline('text-generation', model='distilgpt2')
for t in (0.3, 1.2):
    print(t, gen('Generative AI is', max_new_tokens=25, do_sample=True, temperature=t, top_k=50)[0]['generated_text'])
```

## 3. Inspect the model
Count parameters and compare with the estimate from Unit 2.

```python
model = gen.model
print(sum(p.numel() for p in model.parameters())/1e6, 'M parameters')
print(model.config)
```
